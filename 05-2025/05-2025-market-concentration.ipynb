```python
import pandas as pd
import numpy as np
import yfinance as yf
from datetime import date
from market import MarketData
```

*Data*

```python
start_date = date(2022,1,1)
end_date = date(2024,1,1)

tickers = [
    'XLY',  # Consumer Discretionary
    'XLP',  # Consumer Staples
    'XLE',  # Energy
    'XLF',  # Financials
    'XLV',  # Health Care
    'XLI',  # Industrials
    'XLB',  # Materials
    'XLRE', # Real Estate
    'XLK',  # Technology
    'XLC',  # Communication Services
    'XLU'   # Utilities
]

data = yf.download(tickers,start = start_date, end = end_date)['Close']
```

```python
cov_matrix = (data.pct_change().corr()).values[:3,:3]
```

```python
# Define 3 industries with specific characteristics
industries = {
    'Tech': {'return': 0.12, 'corr_with_finance': 0.4},
    'Finance': {'return': 0.08, 'corr_with_finance': 1.0},
    'Utilities': {'return': 0.04, 'corr_with_finance': 0.1}
}

# Build expected returns vector
expected_returns = np.array([
    industries['Tech']['return'],
    industries['Finance']['return'],
    industries['Utilities']['return']
])


# Initialize and run simulation
sim = MarketData(n_industries=3, n_stocks=500)
sim.assign_stocks_to_industries()
sim.set_industry_characteristics(expected_returns, cov_matrix)
prices = sim.simulate()

# Verify industry performance matches targets
realized_returns = sim.get_industry_performance()
print("Target vs Realized Returns:")
for i, (target, realized) in enumerate(zip(expected_returns, realized_returns)):
    print(f"Industry {i}: Target={target:.2%}, Realized={realized:.2%}")
```
